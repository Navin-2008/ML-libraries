1. Feature Scaling: A technique used to rescale feature values from different measurements into a single uniform scale across a dataset.
2. Standardization (Standard Scaler): Adjusts each feature so that it has a mean ($\mu$) of $0$ and a standard deviation ($\sigma$) of $1$, scaling the data according to a standard normal distribution using the formula $Z = \frac{x - \mu}{\sigma}$.
3. Normalization (Min-Max Scaler): Rescales feature values so they fall within a specific range of $0$ to $1$ using the formula $X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$.

```python
# Reload the original data to recover the clean state
df = pd.read_csv("insurance.csv")
df = df.drop_duplicates()

# Drop rows where target column 'charges' is missing
df = df.dropna(subset=["charges"])

# Fill missing values for numerical columns correctly
df["age"] = df["age"].fillna(df["age"].mean())
df["bmi"] = df["bmi"].fillna(df["bmi"].mean())
df["children"] = df["children"].fillna(df["children"].median())

# Fill missing values for categorical columns using their mode
df["sex"] = df["sex"].fillna(df["sex"].mode()[0])
df["smoker"] = df["smoker"].fillna(df["smoker"].mode()[0])
df["region"] = df["region"].fillna(df["region"].mode()[0])
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["sex","smoker","region"],
    dtype= int
)

df_encoded.head()
```

```python
X_df = df_encoded.drop("charges",axis=1)
y_df = df_encoded["charges"]
print(X_df.shape)
print(y_df.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train_df,X_test_df,y_train_df, y_test_df = train_test_split(X_df,y_df,train_size=0.8,test_size=0.2,random_state=42)
```

```python
print(X_train_df.shape[0])
print(X_test_df.shape[0])
```

```python
print(y_train_df.shape[0])
print(y_test_df.shape[0])
```

```python
from sklearn.preprocessing import StandardScaler

df_scaler = StandardScaler()
scale_columns = ["age", "bmi", "children"]
X_train_df_scaled = X_train_df.copy()
X_test_df_scaled = X_test_df.copy()
X_train_df_scaled[scale_columns] = df_scaler.fit_transform(X_train_df[scale_columns])
X_test_df_scaled[scale_columns] = df_scaler.transform(X_test_df[scale_columns])
```

```python
print(X_train_df_scaled)
print(X_test_df_scaled)
```

Model selection

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

linear_model = LinearRegression()

#Linear model regression - model selection
```

Train the model

```python
linear_model.fit(X_train_df_scaled,y_train_df)
```

Predict  test data

```python
pred = linear_model.predict(X_test_df_scaled)
pred
```

Compare actual and predicted values

```python
df_result = pd.DataFrame({"Actual Charges":y_test_df.values,"Predicted Charges":pred})
df_result.head(10)
```

```python
print(y_test_df-pred)
```

Metrics - Model Evaluation

1. RMSE
2. MSE
3. MAE
4. R2_score

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mse = mean_squared_error(y_test_df,pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_df,pred)
r2 = r2_score(y_test_df,pred)

print("MAE:",mse)
print("RMSE:",rmse)
print("MAE:",mae)
print("R2_score:",r2)
```

Decision Trees 🌳
Definition:

A Decision Tree is a tree-shaped algorithm used to determine a course of action.

Key points:
1. It follows a tree-like structure.
2. Each branch represents a possible decision, occurrence, or reaction.
3. It makes decisions based on conditions.
4. It is used for both classification and regression.
Example:

Predict whether a person will buy insurance based on age, income, and other factors.

Remember for exams: A Decision Tree makes predictions by following a series of conditions, just like a flowchart.
