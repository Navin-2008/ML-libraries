### Creating and Cleaning a Student Dataset
Let's generate a structured DataFrame representing student records, including some missing or invalid data, and then clean it.

```python
import pandas as pd
import numpy as np

# Create a dictionary with simulated student data containing some null/invalid entries
student_data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', np.nan, 'Eva'],
    'Age': [20, 21, np.nan, 19, 22, -5], # -5 is an invalid age
    'Score': [85, np.nan, 90, 78, 88, 95]
}

# Convert to DataFrame
df = pd.DataFrame(student_data)
print("Original Dataset with Issues:")
display(df)
```

```python
# 1. Handle missing names by filling them or dropping the row
# Let's drop rows where 'Name' is missing
df_clean = df.dropna(subset=['Name']).copy()

# 2. Handle missing ages by replacing them with the median age
median_age = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(median_age)

# 3. Correct invalid values (e.g. negative ages)
df_clean.loc[df_clean['Age'] <= 0, 'Age'] = median_age

# 4. Handle missing Scores by replacing them with the average score of the class
mean_score = df_clean['Score'].mean()
df_clean['Score'] = df_clean['Score'].fillna(mean_score)

print("Cleaned Student Dataset:")
display(df_clean)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configure plot style
sns.set_theme(style="whitegrid")

# Create a regression plot
plt.figure(figsize=(8, 6))
sns.regplot(data=df_clean, x='Age', y='Score', color='teal', marker='o', scatter_kws={'s': 80})

# Set title and labels
plt.title('Regression Plot of Student Age vs Score', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Score', fontsize=12)

# Display the plot
plt.show()
```
